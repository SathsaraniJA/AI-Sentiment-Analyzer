```python
# 🧠 AI Sentiment Analyzer Project
# 👩‍💻 Created by: [Your Name]
# 🎯 Goal: Detect whether a text is Positive, Negative, or Neutral using AI

# Step 1: Install Required Libraries
!pip install transformers torch

# Step 2: Import the Hugging Face pipeline
from transformers import pipeline

# Step 3: Load the advanced sentiment analysis model (3 emotions)
sentiment_analyzer = pipeline("sentiment-analysis",
                              model="cardiffnlp/twitter-roberta-base-sentiment")

print("✅ AI Sentiment Analyzer (3-Class Model) is ready to use!\n")

# Step 4: Take user input and analyze sentiment
while True:
    text = input("Enter a sentence (or type 'exit' to stop): ")

    if text.lower() == 'exit':
        print("👋 Thanks for using the Sentiment Analyzer! Goodbye!")
        break

    result = sentiment_analyzer(text)
    label = result[0]['label']
    confidence = result[0]['score']

    # Convert model labels (LABEL_0, LABEL_1, LABEL_2) to actual emotions
    if label == "LABEL_0":
        sentiment = "NEGATIVE"
        emoji = "😡"
    elif label == "LABEL_1":
        sentiment = "NEUTRAL"
        emoji = "😐"
    else:
        sentiment = "POSITIVE"
        emoji = "😊"

    print(f"📝 Text: {text}")
    print(f"💡 Sentiment: {sentiment} {emoji}")
    print(f"📊 Confidence: {confidence:.2f}\n")
```
